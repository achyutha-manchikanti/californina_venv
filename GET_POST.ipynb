```python
import requests

```

**GET Method**

```python
# no parameters no args
url1="http://127.0.0.1:5000"
requests.get(url=url1)
# 200 means success
```

```python
url1="http://127.0.0.1:5000"
response=requests.get(url=url1)
print(response.status_code) # work/fail
print(response.text) # what inside that
```

**POST Method**

```python
# no parameters no args
url2="http://127.0.0.1:5000/add1"
response=requests.post(url=url2)
print(response.status_code)
print(response.text)
```

```python
# no parameters no args
url3="http://127.0.0.1:5000/add2/10/20"
response=requests.get(url=url3)
print(response.status_code)
print(response.text)
```

```python
# no parameters no args
url4="http://127.0.0.1:5000/add3/10/20"
response=requests.post(url=url4)
print(response.status_code)
print(response.text)
```

```python
url5="http://127.0.0.1:5000/add4"
#request.args.get which means we are requesting the arguments
d={'number1':10,'number2':20}
response=requests.get(url=url5,params=d)
print(response.status_code)
print(response.text)
```
